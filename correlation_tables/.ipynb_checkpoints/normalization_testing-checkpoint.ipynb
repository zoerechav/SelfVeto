```python
def one_dimensional_normalization(spline):
    spline = np.round(spline, decimals=8)
    threshold = 1e-8

    spline[spline < threshold] = 0
    dims_nu = spline.shape[0]
    dims_mu = spline.shape[1]
    norm_hist = np.empty((dims_nu,dims_mu))
    for index in range(dims_nu):
        piece = spline[index,:]
        norm_piece = piece /np.sum(piece)
        norm_piece =np.nan_to_num(norm_piece)
        norm_hist[index,:]=norm_piece

    return norm_hist
```

```python
TWO_D = np.load('/data/user/zrechav/output_SelfVeto_Correlation_Tables/histograming/2D/NuMu_Double_Zen_0.8_Depth_2.1.npy')
def two_dimensional_normalization(spline):
   
    spline = np.round(spline, decimals=8)
    threshold = 1e-8
    spline[spline < threshold] = 0
    
    dims_mu = spline.shape[0]
    dims = spline.shape[2]
    norm_2dhist = np.empty((dims_mu, dims_mu, dims))

    for index in range(dims):
        two_d_piece = spline[:, :, index]
        Sum = np.sum(two_d_piece)
        normalized_two_d_piece = two_d_piece / Sum
        normalized_two_d_piece = np.nan_to_num(normalized_two_d_piece)

        norm_2dhist[:, :, index] = normalized_two_d_piece
    return norm_2dhist
    
```

```python
def three_dimensional_normalization(spline):
   
    spline = np.round(spline, decimals=8)
    threshold = 1e-8
    spline[spline < threshold] = 0
    
    dims_mu = spline.shape[0]
    dims = spline.shape[3]
    norm_3dhist = np.empty((dims_mu, dims_mu, dims_mu, dims))

    for index in range(dims):
        three_d_piece = spline[:, :,:, index]
        Sum = np.sum(three_d_piece)
        normalized_three_d_piece = three_d_piece / Sum
        normalized_three_d_piece = np.nan_to_num(normalized_three_d_piece)

        norm_3dhist[:, :, :,index] = normalized_three_d_piece
    return norm_3dhist
```

```python
def four_dimensional_normalization(spline):
   
    spline = np.round(spline, decimals=8)
    threshold = 1e-8
    spline[spline < threshold] = 0
    
    dims_mu = spline.shape[0]
    dims = spline.shape[4]
    norm_4dhist = np.empty((dims_mu, dims_mu, dims_mu, dims_mu, dims))

    for index in range(dims):
        four_d_piece = spline[:, :,:,:, index]
        Sum = np.sum(four_d_piece)
        normalized_four_d_piece = four_d_piece / Sum
        normalized_four_d_piece = np.nan_to_num(normalized_four_d_piece)

        norm_4dhist[:, :, :,:,index] = normalized_four_d_piece
    return norm_4dhist
```
